```python
# TUGAS PERTEMUAN 8 - ALGORITMA DAN STRUKTUR DATA
# Nama  : Ai Asri Astiani
# NIM   : 10225007
# Prodi : Teknik Informatika
# Studi Kasus: Sistem Kasir & Antrean Pesanan Kedai Kopi
# ==============================================================================

# ------------------------------------------------------------------------------
# 1. IMPLEMENTASI ARRAY (1 DIMENSI & 2 DIMENSI)
# Digunakan untuk daftar harga menu dan keranjang belanja (bisa ditambah/diubah)
# ------------------------------------------------------------------------------
print("=== 1. IMPLEMENTASI ARRAY (MENU & KERANJANG PESANAN) ===")

# A. Array 1 Dimensi (1 baris data harga minuman)
daftar_harga = [15000, 18000, 20000, 12000]
print(f"Array 1D (Harga Awal)      : {daftar_harga}")

# Manipulasi Array 1D (Mengubah harga ke-0 dan menambah menu baru)
daftar_harga[0] = 16000
daftar_harga.append(22000)
print(f"Array 1D (Setelah Diubah)  : {daftar_harga}\n")

# B. Array 2 Dimensi (Beberapa baris & kolom: [Nama Menu, Harga, Jumlah Beli])
keranjang_pesanan = [
    ["Kopi Susu Gula Aren", 18000, 2],
    ["Matcha Latte", 20000, 1],
    ["Roti Bakar Coklat", 15000, 2],
    ["Kentang Goreng", 16000, 1]
]

print("Array 2D (Isi Keranjang Pesanan):")
total_bayar = 0
for i in range(len(keranjang_pesanan)):
    menu = keranjang_pesanan[i][0]
    harga = keranjang_pesanan[i][1]
    qty = keranjang_pesanan[i][2]
    subtotal = harga * qty
    total_bayar += subtotal
    print(f"  Baris {i} -> {menu:<20} | {qty} x Rp{harga} = Rp{subtotal}")

# Mengubah jumlah beli pada Baris 1 Kolom 2 (Matcha Latte jadi 2 gelas)
keranjang_pesanan[1][2] = 2
total_bayar = sum(item[1] * item[2] for item in keranjang_pesanan)
print(f"Update Array 2D [Baris 1][Kolom 2] -> Total Akhir: Rp{total_bayar}\n")


# ------------------------------------------------------------------------------
# 2. IMPLEMENTASI LINKED LIST (SINGLY LINKED LIST)
# Digunakan untuk sistem antrean pembuatan pesanan oleh barista
# ------------------------------------------------------------------------------
print("=== 2. IMPLEMENTASI SINGLY LINKED LIST (ANTREAN PESANAN) ===")

class Node:
    def _init_(self, no_antrean, nama_pelanggan):
        self.no_antrean = no_antrean
        self.nama_pelanggan = nama_pelanggan
        self.next = None  # Pointer mengarah ke node berikutnya

class AntreanKafe:
    def _init_(self):
        self.head = None  # Titik awal antrean

    # Menambah antrean di belakang (node baru terhubung ke node terakhir)
    def masuk_antrean(self, no_antrean, nama_pelanggan):
        node_baru = Node(no_antrean, nama_pelanggan)
        if self.head is None:
            self.head = node_baru
            return

        sekarang = self.head
        while sekarang.next:
            sekarang = sekarang.next
        sekarang.next = node_baru

    # Menyelesaikan antrean paling depan (head bergeser ke node berikutnya)
    def selesaikan_pesanan(self):
        if self.head is None:
            print("Antrean kosong.")
            return
        selesai = self.head
        self.head = self.head.next
        print(f"Pesanan Selesai : [{selesai.no_antrean} - {selesai.nama_pelanggan}]")

    # Menampilkan jalur pointer dari head sampai akhir
    def tampilkan_antrean(self):
        sekarang = self.head
        urutan = []
        while sekarang:
            urutan.append(f"[{sekarang.no_antrean}: {sekarang.nama_pelanggan}]")
            sekarang = sekarang.next
        print("Jalur Pointer   : " + " -> ".join(urutan) + " -> None")

#
```
