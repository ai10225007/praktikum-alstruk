```python
# Praktikum 1 Mengenali Tipe Data

# Membuat variabel
nama = "Fajar"
umur = 19
berat = 54.5
terdaftar = True

# Menampilkan nilai dan tipe data
print(nama)
print(type(nama))

print(umur)
print(type(umur))

print(berat)
print(type(berat))

print(terdaftar)
print(type(terdaftar))

# Mengubah nilai umur menjadi 20
umur = 20
print(umur)
print(type(umur))
```

```python
# Praktikum 2 Operator Aritmatika

harga = 12500
jumlah = 4

#Menghitung total harga
total_harga = harga * jumlah

#Menghitung sisa pembagian dengan 2.000
sisa = total_harga % 2000

#Menghitung harga setelah diskon Rp5.000
harga_diskon = total_harga - 5000

#Menampilkan hasil
print("Harga per buku: Rp", harga)
print("Jumlah buku:", jumlah)
print("Total harga: Rp", total_harga)
print("Sisa pembagian total harga dengan 2.000: Rp", sisa)
print("Harga setelah diskon: Rp", harga_diskon)
```

```python
# Praktikum 3 Operator Perbandingan & Logika

# membuat variabel
nilai = 82
kehadiran = 90

# mengecek syarat masing-masing
print("Nilai minimal 75:", nilai >= 75)
print("Kehadiran minimal 80:", kehadiran >= 80)

# mengecek kedua syarat sekaligus menggunakan operator and
print("Memenuhi kedua syarat (and):", nilai >= 75 and kehadiran >= 80)

# mengubah nilai menjadi 70
nilai = 70
print("Nilai diubah menjadi 70, memenuhi syarat (and):", nilai >= 75 and kehadiran >= 80)

# mengecek menggunakan operator or
print("Memenuhi salah satu syarat (or):", nilai >= 75 or kehadiran >= 80)
```

```python
# Praktikum 4 Operator Penugasan

# Nilai awal poin
poin = 100
print("Nilai awal poin:", poin)

# Menambahkan 25 poin
poin += 25
print("Setelah ditambah 25 poin:", poin)

# Mengurangi 10 poin
poin -= 10
print("Setelah dikurangi 10 poin:", poin)

# Mengalikan poin dengan 2
poin *= 2
print("Setelah dikalikan 2:", poin)

# Membagi poin dengan 2
poin /= 2
print("Setelah dibagi 2:", poin)
```

```python
# Praktikum 5 Struktur Data Dasar

# Membuat struktur data
nilai = [78, 85, 90, 85]
mahasiswa = {"nama": "Nadia", "kelas": "IF-A"}
koordinat = (5, 8)
angka_unik = {2, 4, 4, 6, 8}

# Menampilkan nilai pertama dari list
print("Nilai pertama dari list:", nilai[0])

# Menambahkan angka 95 ke dalam list
nilai.append(95)
print("List setelah ditambah angka 95:", nilai)

# Menampilkan nama mahasiswa dari dictionary
print("Nama mahasiswa:", mahasiswa["nama"])

# Menampilkan koordinat pertama dari tuple
print("Koordinat pertama:", koordinat[0])

# Menampilkan angka unik yang sudah diurutkan
print("Angka unik yang terurut:", sorted(angka_unik))
```

```python
# Tantangan Terintegrasi - Program Rekap Nilai

# Data Mahasiswa
nama = "Ai Asri Astiani"
nilai_tugas = 90
nilai_ujian = 92

# Menghitung rata-rata
rata_rata = (nilai_tugas + nilai_ujian) / 2

# Memeriksa status kelulusan
lulus = rata_rata >= 75

# Menampilkan hasil
print("=== PROGRAM REKAP NILAI ===")
print("Nama mahasiswa:", nama)
print("Nilai tugas:", nilai_tugas)
print("Nilai ujian:", nilai_ujian)
print("Rata-rata:", rata_rata)
print("Status lulus:", lulus)

# Pengujian kedua dengan nilai berbeda
print("\n=== PENGUJIAN KEDUA ===")

nama = "Ai Asri Astiani"
nilai_tugas = 85
nilai_ujian = 77

rata_rata = (nilai_tugas + nilai_ujian) / 2
lulus = rata_rata >= 75

print("Nama mahasiswa:", nama)
print("Nilai tugas:", nilai_tugas)
print("Nilai ujian:", nilai_ujian)
print("Rata-rata:", rata_rata)
print("Status lulus:", lulus)
```
