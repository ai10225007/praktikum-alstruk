```python
# TUGAS PERTEMUAN 8 - ALGORITMA DAN STRUKTUR DATA
# Nama  : Ai Asri Astiani
# NIM   : 10225007
# Prodi : Teknik Informatika
# Studi Kasus: Sistem Pengelolaan Nilai & Antrean Bimbingan Mahasiswa
# ==============================================================================

# 1. IMPLEMENTASI ARRAY (1 DIMENSI & 2 DIMENSI)
print("=== 1. IMPLEMENTASI ARRAY (DATA NILAI MAHASISWA) ===")

# A. Array 1 Dimensi (Daftar Nilai Kuis)
nilai_kuis = [75, 80, 85, 90]
print(f"Array 1D (Nilai Awal)      : {nilai_kuis}")

# Mengubah elemen pertama & menambah nilai baru
nilai_kuis[0] = 78
nilai_kuis.append(95)
print(f"Array 1D (Setelah Diubah)  : {nilai_kuis}\n")

# B. Array 2 Dimensi (Tabel Nilai: [Nama Mahasiswa, Tugas, UTS, UAS])
data_nilai = [
    ["Ai Asri Astiani", 88, 85, 90],
    ["Nadia Putri", 80, 78, 84],
    ["Rizky Pratama", 75, 82, 80],
    ["Dimas Saputra", 90, 88, 92]
]

print("Array 2D (Rekap Nilai Mahasiswa):")
for i in range(len(data_nilai)):
    nama = data_nilai[i][0]
    tugas = data_nilai[i][1]
    uts = data_nilai[i][2]
    uas = data_nilai[i][3]
    rata_rata = round((tugas + uts + uas) / 3, 2)
    print(f"  Baris {i} -> {nama:<18} | Tugas: {tugas}, UTS: {uts}, UAS: {uas} | Rata-rata: {rata_rata}")

# Update nilai UTS milik Nadia Putri (Baris ke-1, Kolom ke-2)
data_nilai[1][2] = 85
rata_ai = round((data_nilai[0][1] + data_nilai[0][2] + data_nilai[0][3]) / 3, 2)
print(f"Update Array 2D [Baris 1][Kolom 2] -> Nilai UTS Nadia menjadi: {data_nilai[1][2]}\n")


# 2. IMPLEMENTASI SINGLY LINKED LIST (ANTREAN BIMBINGAN / KONSULTASI NILAI)
print("=== 2. IMPLEMENTASI SINGLY LINKED LIST (ANTREAN KONSULTASI) ===")

class Node:
    pass

class AntreanBimbingan:
    head = None

    def tambah_antrean(self, nim, nama):
        node_baru = Node()
        node_baru.nim = nim
        node_baru.nama = nama
        node_baru.next = None

        if self.head is None:
            self.head = node_baru
            return
        sekarang = self.head
        while sekarang.next:
            sekarang = sekarang.next
        sekarang.next = node_baru

    def panggil_mahasiswa(self):
        if self.head is None:
            print("Antrean kosong.")
            return
        dipanggil = self.head
        self.head = self.head.next
        print(f"Selesai Konsultasi : [{dipanggil.nim} - {dipanggil.nama}]")

    def tampilkan_antrean(self):
        sekarang = self.head
        daftar = []
        while sekarang:
            daftar.append(f"[{sekarang.nim}: {sekarang.nama}]")
            sekarang = sekarang.next
        print("Jalur Pointer      : " + " -> ".join(daftar) + " -> None")

antrean = AntreanBimbingan()
antrean.tambah_antrean("10225007", "Ai Asri Astiani")
antrean.tambah_antrean("10225008", "Nadia Putri")
antrean.tambah_antrean("10225009", "Rizky Pratama")
antrean.tambah_antrean("10225010", "Dimas Saputra")

antrean.tampilkan_antrean()
antrean.panggil_mahasiswa()
antrean.tampilkan_antrean()
print()


# 3. IMPLEMENTASI TUPLE (IDENTITAS MAHASISWA & TRANSKRIP PERMANEN)
print("=== 3. IMPLEMENTASI TUPLE (IDENTITAS & HASIL AKHIR PERMANEN) ===")

identitas_mhs = ("10225007", "Ai Asri Astiani", "Teknik Informatika", "Semester 2")
predikat_kelas = ("A", "B", "B", "A", "A")
transkrip_final = (identitas_mhs[0], identitas_mhs[1], rata_ai, "LULUS")

print(f"Identitas Mahasiswa : {identitas_mhs[1]} ({identitas_mhs[0]}) - {identitas_mhs[2]} ({identitas_mhs[3]})")
print(f"Daftar Predikat     : {predikat_kelas} | Total (len): {len(predikat_kelas)} | Predikat 'A' (count): {predikat_kelas.count('A')}")
print(f"Transkrip Final     : {transkrip_final} -> Status: Permanen (Tidak dapat diubah)")
```
