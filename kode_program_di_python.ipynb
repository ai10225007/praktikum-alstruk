```python
# Program Pemeriksaan Izin Mengendarai Sepeda Motor

print("=== SISTEM PEMERIKSAAN IZIN MENGENDARAI MOTOR ===")

# Input data pengendara
nama = input("Masukkan nama pengendara: ")
usia = int(input("Masukkan usia (tahun): "))

# Penerapan Struktur Percabangan (if-else)
if usia >= 17:
    status = "DIPERBOLEHKAN"
    keterangan = "Sudah memenuhi syarat usia minimal (Memiliki/Boleh membuat SIM)."
else:
    status = "DILARANG (DITILANG)"
    keterangan = "Anak di bawah umur! Belum memenuhi syarat usia minimal 17 tahun."

# Menampilkan Hasil Evaluasi
print("\n--- HASIL PEMERIKSAAN ---")
print(f"Nama Pengendara : {nama}")
print(f"Usia            : {usia} tahun")
print(f"Status Izin     : {status}")
print(f"Keterangan      : {keterangan}")
```
